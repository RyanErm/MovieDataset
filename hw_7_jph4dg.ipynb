## Our solution! A new algorithm for Netflix should satisfies customers concerns that they are not being provided a wide variety of content to watch. To that end, our algorithm will not only draw from the users actions, but will also reccomend shows outside of their "comfort zone" (usual watching habits). For example, a user who has been watching many documentaries on war could be reccomended a fictional political thriller. By providing new generes of content, users hopefully find whole new fields of content to watch, and remain engaged on Netflix's platform.
